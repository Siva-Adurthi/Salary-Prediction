```python
X=data.drop(columns=['salary'])
y=data['salary']
y_log=np.log1p(y)
```

```python
high=['job_title','industry','location']
low=['education_level','company_size','remote_work']
numerical=['experience_years','skills_count','certifications']
```

```python
preprocessor=ColumnTransformer(transformers=[('num',StandardScaler(),numerical),('high_values',TargetEncoder(),high),('low_values',OneHotEncoder(handle_unknown='ignore',sparse_output=False),low)])
```

```python
model=Pipeline(steps=[('preprocessor',preprocessor),('regressor',RandomForestRegressor(n_estimators=100,max_depth=15,random_state=42,n_jobs=-1))])
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y_log,test_size=0.2,random_state=42)
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred_log=np.expm1(y_pred)
y_original=np.expm1(y_test)
```

```python
score=r2_score(y_original,y_pred_log)
score
```
